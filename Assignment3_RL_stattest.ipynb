```python
import pandas as pd
from scipy import stats

df_benchmark = pd.read_csv("Users/lukachkadua/experiment_results_benchmark.csv")
df_benchmark
```

```python

def load_avgs(path):
    df = pd.read_csv(path)
    avgs_list = []
    for col in df.columns:
        if col.startswith("Avg_"):
            value = df[col].iloc[0]
            avgs_list.append(value)

    return avgs_list


benchmark = load_avgs("Users/lukachkadua/experiment_results_benchmark.csv")


```

```python
uniform = load_avgs("Users/lukachkadua/experiment_results_uniform.csv")
linear = load_avgs("Users/lukachkadua/experiment_results_linear.csv")
decay = load_avgs("Users/lukachkadua/experiment_results_decay.csv")

# strategies vs benchmark
print("uniform vs Benchmark:", stats.ttest_rel(uniform, benchmark))
print("linear vs Benchmark:", stats.ttest_rel(linear, benchmark))
print("decay vs Benchmark:", stats.ttest_rel(decay, benchmark))
```

```python
# strategies against each other 

print("uniform vs linear:", stats.ttest_rel(uniform, linear))
print("uniform vs decay: ", stats.ttest_rel(uniform, decay))
print("linear vs decay:  ", stats.ttest_rel(linear, decay))
```
